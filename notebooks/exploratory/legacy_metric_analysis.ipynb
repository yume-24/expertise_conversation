```python
#one way anova

def eta_squared_oneway(feature, group_col="expertise_B"):
    d = df[[feature, group_col]].dropna()
    groups = [g[feature].values for _, g in d.groupby(group_col)]
    if len(groups) < 2:
        return np.nan

    y = d[feature].values
    grand_mean = np.mean(y)

    ss_between = 0.0
    for _, g in d.groupby(group_col):
        ss_between += len(g) * (g[feature].mean() - grand_mean) ** 2

    ss_total = np.sum((y - grand_mean) ** 2)
    return ss_between / ss_total if ss_total > 0 else np.nan

eta_rows = []
for col in num_df.columns:
    if col in ["expertise_A_ord", "expertise_B_ord", "expertise_gap_signed", "expertise_gap_abs"]:
        continue
    eta = eta_squared_oneway(col, "expertise_B")
    eta_rows.append((col, eta))

eta_df = pd.DataFrame(eta_rows, columns=["feature", "eta_sq"]).sort_values("eta_sq", ascending=False)
eta_df.head(20)
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# =========================
# 1) EXPERTISE LABELING
# =========================

EXPERTISE_ORDER = ["Child", "Teen", "College Student", "Graduate Student", "Expert"]
LABEL_TO_ORD = {lab: i for i, lab in enumerate(EXPERTISE_ORDER)}

SUFFIX_TO_EXPERTISE = {
    "12": "Child",
    "13": "Teen",
    "14": "College Student",
    "15": "Graduate Student",
    "16": "Expert",
}

EXPERTISE_PATTERNS = [
    (r"\bchild\b", "Child"),
    (r"\bteen(ager)?\b|\bhigh school\b|\byoung adult\b", "Teen"),
    (r"\bcollege\b|\bundergrad\b", "College Student"),
    (r"\bgrad(uate)?\b", "Graduate Student"),
    (r"\bexpert\b|\bphd\b|\bphysicist\b", "Expert"),
]

def infer_expertise_from_speaker_text(text):
    if pd.isna(text):
        return None
    s = str(text)

    # Prefer parentheses content first
    parens = re.findall(r"\(([^)]*)\)", s)
    for chunk in parens:
        chunk_l = chunk.lower()
        for pat, label in EXPERTISE_PATTERNS:
            if re.search(pat, chunk_l):
                return label

    # Fallback: scan whole string
    s_l = s.lower()
    for pat, label in EXPERTISE_PATTERNS:
        if re.search(pat, s_l):
            return label
    return None

def infer_expertise_from_convo_id(convo_id):
    if pd.isna(convo_id):
        return None
    m = re.search(r"_(\d{2})\b", str(convo_id))
    if not m:
        return None
    return SUFFIX_TO_EXPERTISE.get(m.group(1))

def add_expertise_labels(df: pd.DataFrame, assume_speakerA_is_expert=False) -> pd.DataFrame:
    df = df.copy()

    if "speaker_A" not in df.columns or "speaker_B" not in df.columns:
        raise ValueError("Expected columns speaker_A and speaker_B in the CSV.")

    if "convo_id" not in df.columns:
        # still possible to label from speaker text only, but convo_id suffix fallback won't work
        df["convo_id"] = np.nan

    df["expertise_A"] = df["speaker_A"].apply(infer_expertise_from_speaker_text)
    df["expertise_B"] = df["speaker_B"].apply(infer_expertise_from_speaker_text)

    # fallback via convo_id suffix (very useful for Wired Explain sets)
    mask_B_missing = df["expertise_B"].isna()
    df.loc[mask_B_missing, "expertise_B"] = df.loc[mask_B_missing, "convo_id"].apply(infer_expertise_from_convo_id)

    if assume_speakerA_is_expert:
        mask_A_missing = df["expertise_A"].isna()
        df.loc[mask_A_missing, "expertise_A"] = "Expert"

    df["expertise_A_ord"] = df["expertise_A"].map(LABEL_TO_ORD)
    df["expertise_B_ord"] = df["expertise_B"].map(LABEL_TO_ORD)

    # signed gap: B - A (positive means B is "more expert" than A)
    df["expertise_gap_signed"] = df["expertise_B_ord"] - df["expertise_A_ord"]
    df["expertise_gap_abs"] = df["expertise_gap_signed"].abs()

    return df

# =========================
# 2) PARTIAL CORR (RESIDUAL)
# =========================

def _safe_numeric(s: pd.Series) -> pd.Series:
    return pd.to_numeric(s, errors="coerce")

def partial_corr_residual(
    df: pd.DataFrame,
    x: str,
    y: str,
    controls: list[str],
    min_n: int = 8,
) -> tuple[float, int]:
    """
    Partial corr(x, y | controls) by residualizing x and y on controls, then correlating residuals.
    Returns (r, n). r is NaN if not computable.
    """
    cols = [x, y] + list(controls)
    cols = [c for c in cols if c in df.columns]

    if x not in cols or y not in cols:
        return np.nan, 0

    d = df[cols].copy()

    # coerce x and y to numeric; controls handled as numeric as well
    for c in cols:
        d[c] = _safe_numeric(d[c])

    d = d.dropna()
    if len(d) < min_n:
        return np.nan, len(d)

    # controls matrix
    X = d[controls].values if len(controls) > 0 else None

    # if no controls, just corr(x, y)
    if X is None or X.size == 0:
        xv = d[x].values
        yv = d[y].values
        if np.std(xv) == 0 or np.std(yv) == 0:
            return np.nan, len(d)
        return float(np.corrcoef(xv, yv)[0, 1]), len(d)

    # residualize x
    regx = LinearRegression().fit(X, d[x].values)
    rx = d[x].values - regx.predict(X)

    # residualize y
    regy = LinearRegression().fit(X, d[y].values)
    ry = d[y].values - regy.predict(X)

    if np.std(rx) == 0 or np.std(ry) == 0:
        return np.nan, len(d)

    r = float(np.corrcoef(rx, ry)[0, 1])
    return r, len(d)

# =========================
# 3) RUNNER: FEATURE LIST + REPORT
# =========================

DEFAULT_COUNT_COLS = ["n_utterances", "n_points", "n_A", "n_B"]
DEFAULT_ALIGNMENT_COLS = [
    # use whichever exist in your file
    "align_eps",
    "align_lag_A_to_B",
    "align_lag_B_to_A",
    "align_lag_asymmetry_AminusB",
    "influence_A_toward_B",
    "influence_B_toward_A",
    "influence_asymmetry_AminusB",
]

def get_numeric_feature_cols(df: pd.DataFrame, exclude: set[str]) -> list[str]:
    numeric_cols = []
    for c in df.columns:
        if c in exclude:
            continue
        # keep numeric-like cols; if object col but numeric strings, we'll handle later in coercion
        numeric_cols.append(c)
    return numeric_cols

def run_partial_corr_suite(
    df: pd.DataFrame,
    *,
    target_col: str,
    controls: list[str],
    exclude_cols: list[str] = None,
    min_n: int = 8,
    title: str = "",
) -> pd.DataFrame:
    """
    Partial corr(feature, target | controls) for all features.
    """
    exclude_cols = exclude_cols or []
    exclude = set(exclude_cols + [target_col] + controls)

    feature_cols = get_numeric_feature_cols(df, exclude=exclude)

    rows = []
    for feat in feature_cols:
        r, n = partial_corr_residual(df, x=feat, y=target_col, controls=controls, min_n=min_n)
        rows.append((feat, r, n))

    out = pd.DataFrame(rows, columns=["feature", "partial_r", "n"])
    out["abs_partial_r"] = out["partial_r"].abs()
    out = out.sort_values("abs_partial_r", ascending=False)

    if title:
        print(f"\n=== {title} ===")
        print(out.head(15).to_string(index=False))

    return out

# =========================
# 4) LOAD + LABEL + RUN BOTH FILES
# =========================

def analyze_file(
    path: str,
    *,
    assume_speakerA_is_expert: bool = False,
    min_n: int = 8,
    save_prefix: str = None,
):
    df = pd.read_csv(path)
    df_labeled = add_expertise_labels(df, assume_speakerA_is_expert=assume_speakerA_is_expert)

    # Choose controls you *actually* have
    controls_gap = []
    # recommended: control for counts + (optionally) alignment metrics if you want "beyond alignment"
    for c in DEFAULT_COUNT_COLS + DEFAULT_ALIGNMENT_COLS:
        if c in df_labeled.columns:
            controls_gap.append(c)

    # Partial corr: feature vs expertise_gap_abs
    pc_gap_abs = run_partial_corr_suite(
        df_labeled,
        target_col="expertise_gap_abs",
        controls=controls_gap,
        exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
        min_n=min_n,
        title=f"{path} | partial corr with expertise_gap_abs (controls={controls_gap})",
    )

    # Partial corr: feature vs expertise_gap_signed
    pc_gap_signed = run_partial_corr_suite(
        df_labeled,
        target_col="expertise_gap_signed",
        controls=controls_gap,
        exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
        min_n=min_n,
        title=f"{path} | partial corr with expertise_gap_signed (controls={controls_gap})",
    )

    # OPTIONAL: if you want "what predicts alignment, beyond gap?"
    # Example target: align_lag_asymmetry_AminusB (if present)
    pc_align = None
    align_target = "align_lag_asymmetry_AminusB"
    if align_target in df_labeled.columns:
        controls_align = [c for c in (DEFAULT_COUNT_COLS + ["expertise_gap_abs", "expertise_gap_signed"]) if c in df_labeled.columns]
        pc_align = run_partial_corr_suite(
            df_labeled,
            target_col=align_target,
            controls=controls_align,
            exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
            min_n=min_n,
            title=f"{path} | partial corr with {align_target} (controls={controls_align})",
        )

    if save_prefix:
        df_labeled.to_csv(f"{save_prefix}_labeled.csv", index=False)
        pc_gap_abs.to_csv(f"{save_prefix}_pc_gap_abs.csv", index=False)
        pc_gap_signed.to_csv(f"{save_prefix}_pc_gap_signed.csv", index=False)
        if pc_align is not None:
            pc_align.to_csv(f"{save_prefix}_pc_align.csv", index=False)

    return df_labeled, pc_gap_abs, pc_gap_signed, pc_align

# =========================
# 5) RUN THESE (EDIT PATHS)
# =========================

# Example paths — replace with your actual notebook paths:
SENT_PATH = "first_approach/all_metrics.csv"
TRI_PATH  = "first_approach/all_metrics_trigram.csv"

df_sent_labeled, sent_pc_abs, sent_pc_signed, sent_pc_align = analyze_file(
    SENT_PATH,
    assume_speakerA_is_expert=False,   # set True only if you want to force missing A labels -> Expert
    min_n=8,
    save_prefix="sentence",
)

df_tri_labeled, tri_pc_abs, tri_pc_signed, tri_pc_align = analyze_file(
    TRI_PATH,
    assume_speakerA_is_expert=False,
    min_n=8,
    save_prefix="trigram",
)

print("\nDone. Saved labeled + partial correlation tables if save_prefix was provided.")
```

```python
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# =========================
# 1) EXPERTISE LABELING (HOST=A=Expert, GUEST=B=convo suffix)
# =========================

EXPERTISE_ORDER = ["Child", "Teen", "College Student", "Graduate Student", "Expert"]
LABEL_TO_ORD = {lab: i for i, lab in enumerate(EXPERTISE_ORDER)}

SUFFIX_TO_EXPERTISE = {
    "12": "Child",
    "13": "Teen",
    "14": "College Student",
    "15": "Graduate Student",
    "16": "Expert",
}

def infer_guest_expertise_from_convo_id(convo_id: str):
    if pd.isna(convo_id):
        return None
    m = re.search(r"_(\d{2})\b", str(convo_id))
    if not m:
        return None
    return SUFFIX_TO_EXPERTISE.get(m.group(1))

def add_expertise_labels_host_guest(df: pd.DataFrame) -> pd.DataFrame:
    """
    Assumptions:
      - speaker_A is always host Expert
      - speaker_B is always guest, expertise determined by convo_id suffix _12.._16
    """
    df = df.copy()

    if "convo_id" not in df.columns:
        raise ValueError("Expected a 'convo_id' column to infer guest expertise from suffix (_12.._16).")
    if "speaker_A" not in df.columns or "speaker_B" not in df.columns:
        raise ValueError("Expected columns 'speaker_A' and 'speaker_B'.")

    # Host (A) is always Expert
    df["expertise_A"] = "Expert"
    df["expertise_B"] = df["convo_id"].apply(infer_guest_expertise_from_convo_id)

    # ordinals
    df["expertise_A_ord"] = df["expertise_A"].map(LABEL_TO_ORD)   # should always be 4
    df["expertise_B_ord"] = df["expertise_B"].map(LABEL_TO_ORD)

    # gaps
    # signed: guest - host (so usually negative unless guest is Expert)
    df["expertise_gap_signed"] = df["expertise_B_ord"] - df["expertise_A_ord"]
    df["expertise_gap_abs"] = df["expertise_gap_signed"].abs()

    # quick sanity checks
    if df["expertise_B"].isna().any():
        missing = df.loc[df["expertise_B"].isna(), "convo_id"].head(10).tolist()
        print("[WARN] Some rows could not infer guest expertise from convo_id suffix. Example convo_id:", missing)

    return df


# =========================
# 2) PARTIAL CORR (RESIDUAL)
# =========================

def _safe_numeric(s: pd.Series) -> pd.Series:
    return pd.to_numeric(s, errors="coerce")

def partial_corr_residual(
    df: pd.DataFrame,
    x: str,
    y: str,
    controls: list[str],
    min_n: int = 8,
) -> tuple[float, int]:
    cols = [x, y] + list(controls)

    # If any required column missing, bail
    for c in [x, y] + list(controls):
        if c not in df.columns:
            return np.nan, 0

    d = df[cols].copy()

    # coerce everything to numeric
    for c in cols:
        d[c] = _safe_numeric(d[c])

    d = d.dropna()
    if len(d) < min_n:
        return np.nan, len(d)

    X = d[controls].values if len(controls) > 0 else None

    # no controls -> Pearson corr
    if X is None or X.size == 0:
        xv = d[x].values
        yv = d[y].values
        if np.std(xv) == 0 or np.std(yv) == 0:
            return np.nan, len(d)
        return float(np.corrcoef(xv, yv)[0, 1]), len(d)

    # residualize x
    regx = LinearRegression().fit(X, d[x].values)
    rx = d[x].values - regx.predict(X)

    # residualize y
    regy = LinearRegression().fit(X, d[y].values)
    ry = d[y].values - regy.predict(X)

    if np.std(rx) == 0 or np.std(ry) == 0:
        return np.nan, len(d)

    return float(np.corrcoef(rx, ry)[0, 1]), len(d)


# =========================
# 3) RUNNER: FEATURE LIST + REPORT
# =========================

DEFAULT_COUNT_COLS = ["n_utterances", "n_points", "n_A", "n_B"]
DEFAULT_ALIGNMENT_COLS = [
    "align_eps",
    "align_lag_A_to_B",
    "align_lag_B_to_A",
    "align_lag_asymmetry_AminusB",
    "influence_A_toward_B",
    "influence_B_toward_A",
    "influence_asymmetry_AminusB",
]

def get_feature_cols(df: pd.DataFrame, exclude: set[str]) -> list[str]:
    # we'll coerce to numeric later; keep anything not excluded
    return [c for c in df.columns if c not in exclude]

def run_partial_corr_suite(
    df: pd.DataFrame,
    *,
    target_col: str,
    controls: list[str],
    exclude_cols: list[str] = None,
    min_n: int = 8,
    title: str = "",
) -> pd.DataFrame:
    exclude_cols = exclude_cols or []
    exclude = set(exclude_cols + [target_col] + controls)

    feature_cols = get_feature_cols(df, exclude=exclude)

    rows = []
    for feat in feature_cols:
        r, n = partial_corr_residual(df, x=feat, y=target_col, controls=controls, min_n=min_n)
        rows.append((feat, r, n))

    out = pd.DataFrame(rows, columns=["feature", "partial_r", "n"])
    out["abs_partial_r"] = out["partial_r"].abs()
    out = out.sort_values("abs_partial_r", ascending=False)

    if title:
        print(f"\n=== {title} ===")
        print(out.head(15).to_string(index=False))

    return out


# =========================
# 4) LOAD + LABEL + RUN
# =========================

def analyze_file_host_guest(
    path: str,
    *,
    min_n: int = 8,
    save_prefix: str | None = None,
):
    df = pd.read_csv(path)

    # apply host/guest labeling assumptions
    df_labeled = add_expertise_labels_host_guest(df)

    # controls (only keep ones that exist)
    controls_gap = [c for c in (DEFAULT_COUNT_COLS + DEFAULT_ALIGNMENT_COLS) if c in df_labeled.columns]

    # partial corr vs abs gap
    pc_gap_abs = run_partial_corr_suite(
        df_labeled,
        target_col="expertise_gap_abs",
        controls=controls_gap,
        exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
        min_n=min_n,
        title=f"{path} | partial corr with expertise_gap_abs (controls={controls_gap})",
    )

    # partial corr vs signed gap
    pc_gap_signed = run_partial_corr_suite(
        df_labeled,
        target_col="expertise_gap_signed",
        controls=controls_gap,
        exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
        min_n=min_n,
        title=f"{path} | partial corr with expertise_gap_signed (controls={controls_gap})",
    )

    # optional: alignment target
    pc_align = None
    align_target = "align_lag_asymmetry_AminusB"
    if align_target in df_labeled.columns:
        controls_align = [c for c in (DEFAULT_COUNT_COLS + ["expertise_gap_abs", "expertise_gap_signed"]) if c in df_labeled.columns]
        pc_align = run_partial_corr_suite(
            df_labeled,
            target_col=align_target,
            controls=controls_align,
            exclude_cols=["convo_id", "speaker_A", "speaker_B", "expertise_A", "expertise_B"],
            min_n=min_n,
            title=f"{path} | partial corr with {align_target} (controls={controls_align})",
        )

    if save_prefix:
        df_labeled.to_csv(f"{save_prefix}_labeled.csv", index=False)
        pc_gap_abs.to_csv(f"{save_prefix}_pc_gap_abs.csv", index=False)
        pc_gap_signed.to_csv(f"{save_prefix}_pc_gap_signed.csv", index=False)
        if pc_align is not None:
            pc_align.to_csv(f"{save_prefix}_pc_align.csv", index=False)

    return df_labeled, pc_gap_abs, pc_gap_signed, pc_align


# =========================
# 5) RUN
# =========================

SENT_PATH = "first_approach/all_conversations_metrics_speaker1A_new.csv"
TRI_PATH  = "first_approach/all_metrics_trigram.csv"

df_sent_labeled, sent_pc_abs, sent_pc_signed, sent_pc_align = analyze_file_host_guest(
    SENT_PATH, min_n=8, save_prefix="sentence"
)

df_tri_labeled, tri_pc_abs, tri_pc_signed, tri_pc_align = analyze_file_host_guest(
    TRI_PATH, min_n=8, save_prefix="trigram"
)

print("\nDone.")
```

```python
#helpers
import re
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from scipy import stats

# -------------------------
# Expertise labeling (A=host expert, B=guest by convo suffix)
# -------------------------
EXPERTISE_ORDER = ["Child", "Teen", "College Student", "Graduate Student", "Expert"]
LABEL_TO_ORD = {lab: i for i, lab in enumerate(EXPERTISE_ORDER)}

SUFFIX_TO_EXPERTISE = {
    "12": "Child",
    "13": "Teen",
    "14": "College Student",
    "15": "Graduate Student",
    "16": "Expert",
}

def infer_guest_expertise_from_convo_id(convo_id: str):
    if pd.isna(convo_id):
        return None
    m = re.search(r"_(\d{2})\b", str(convo_id))
    if not m:
        return None
    return SUFFIX_TO_EXPERTISE.get(m.group(1))

def add_expertise_labels_host_guest(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    if "convo_id" not in df.columns:
        raise ValueError("Expected 'convo_id' to infer guest expertise from suffix (_12.._16).")

    df["expertise_A"] = "Expert"
    df["expertise_B"] = df["convo_id"].apply(infer_guest_expertise_from_convo_id)

    df["expertise_A_ord"] = df["expertise_A"].map(LABEL_TO_ORD)   # always 4
    df["expertise_B_ord"] = df["expertise_B"].map(LABEL_TO_ORD)

    df["expertise_gap_signed"] = df["expertise_B_ord"] - df["expertise_A_ord"]  # usually negative
    df["expertise_gap_abs"] = df["expertise_gap_signed"].abs()

    if df["expertise_B"].isna().any():
        bad = df.loc[df["expertise_B"].isna(), "convo_id"].head(10).tolist()
        print("[WARN] Could not infer guest expertise for some rows. Example convo_id:", bad)

    return df

# -------------------------
# Partial correlation + p-value
# -------------------------
def _safe_numeric(s: pd.Series) -> pd.Series:
    return pd.to_numeric(s, errors="coerce")

def _residualize(y: np.ndarray, X: np.ndarray) -> np.ndarray:
    """Return residuals of y ~ X (with intercept)."""
    if X.ndim == 1:
        X = X.reshape(-1, 1)
    reg = LinearRegression().fit(X, y)
    return y - reg.predict(X)

def partial_corr_and_p(df: pd.DataFrame, x: str, y: str, controls: list[str], min_n: int = 8):
    """
    Partial correlation via residualizing x and y on controls.
    Returns dict with r, p, n, df (degrees of freedom), k (# controls).
    """
    cols = [x, y] + list(controls)
    for c in cols:
        if c not in df.columns:
            return {"r": np.nan, "p": np.nan, "n": 0, "df": np.nan, "k": len(controls)}

    d = df[cols].copy()
    for c in cols:
        d[c] = _safe_numeric(d[c])
    d = d.dropna()

    n = len(d)
    k = len(controls)
    if n < max(min_n, k + 3):  # need enough degrees of freedom
        return {"r": np.nan, "p": np.nan, "n": n, "df": np.nan, "k": k}

    xv = d[x].to_numpy()
    yv = d[y].to_numpy()

    if k == 0:
        # simple Pearson
        if np.std(xv) == 0 or np.std(yv) == 0:
            return {"r": np.nan, "p": np.nan, "n": n, "df": n - 2, "k": 0}
        r, p = stats.pearsonr(xv, yv)
        return {"r": float(r), "p": float(p), "n": n, "df": n - 2, "k": 0}

    X = d[controls].to_numpy()
    rx = _residualize(xv, X)
    ry = _residualize(yv, X)

    if np.std(rx) == 0 or np.std(ry) == 0:
        return {"r": np.nan, "p": np.nan, "n": n, "df": n - k - 2, "k": k}

    r = float(np.corrcoef(rx, ry)[0, 1])

    # t-test for partial correlation
    dof = n - k - 2
    if dof <= 0 or abs(r) >= 1:
        p = np.nan
    else:
        t = r * np.sqrt(dof / (1 - r**2))
        p = float(2 * (1 - stats.t.cdf(abs(t), df=dof)))

    return {"r": r, "p": p, "n": n, "df": dof, "k": k}

# -------------------------
# FDR correction (Benjamini–Hochberg)
# -------------------------
def fdr_bh(pvals: np.ndarray):
    """
    Return q-values (BH-FDR). Preserves NaNs.
    """
    p = np.array(pvals, dtype=float)
    q = np.full_like(p, np.nan)

    mask = ~np.isnan(p)
    pv = p[mask]
    m = pv.size
    if m == 0:
        return q

    order = np.argsort(pv)
    ranked = pv[order]
    q_ranked = ranked * m / (np.arange(1, m + 1))

    # monotone non-increasing adjustment when mapping back
    q_ranked = np.minimum.accumulate(q_ranked[::-1])[::-1]
    q_ranked = np.clip(q_ranked, 0, 1)

    # map back
    q_tmp = np.empty_like(pv)
    q_tmp[order] = q_ranked
    q[mask] = q_tmp
    return q
```

```python
#targeted partial corr matrix
# Controls: keep these "mechanical" things as controls;
# DO NOT include the target itself or obvious algebraic transforms of it.
DEFAULT_CONTROLS = [
    "n_utterances", "n_A", "n_B",
    # optionally include "align_eps" if you believe it's a nuisance scale factor
    "align_eps",
]

# Choose columns (key explanatory variables you care about)
COL_VARS = [
    "expertise_gap_abs",
    "expertise_gap_signed",
    "align_lag_asymmetry_AminusB",
    "influence_asymmetry_AminusB",
    "n_utterances",
]

# Choose rows (metrics you want to compare against those explanatory variables)
ROW_METRICS = [
    "centroid_dist_mean",
    "centroid_dist_slope",
    "centroid_dist_std",
    "cross_nn_A_to_B",
    "cross_nn_B_to_A",
    "hausdorff_AB",
    "turn_jump_mean",
    "entropy_total",
    "entropy_A",
    "entropy_B",
    "entropy_slope_A",
    "entropy_slope_B",
    "mi_speaker_cluster",
    "nmi_speaker_cluster",
    "mi_turn_clusters",
    "nmi_turn_clusters",
    "hull_overlap_ratio_min",
    "containment_A_in_B",
    "containment_B_in_A",
]

# Keep only rows/cols that exist
ROW_METRICS = [m for m in ROW_METRICS if m in df.columns]
COL_VARS = [c for c in COL_VARS if c in df.columns]
CONTROLS = [c for c in DEFAULT_CONTROLS if c in df.columns]

results = []
for row_metric in ROW_METRICS:
    for col_var in COL_VARS:
        # controls should NOT include the current x or y
        controls_here = [c for c in CONTROLS if c not in (row_metric, col_var)]
        out = partial_corr_and_p(df, x=row_metric, y=col_var, controls=controls_here, min_n=8)
        results.append({
            "metric": row_metric,
            "target": col_var,
            "partial_r": out["r"],
            "p": out["p"],
            "q_fdr": np.nan,  # fill after
            "n": out["n"],
            "dof": out["df"],
            "k_controls": out["k"],
            "controls_used": ",".join(controls_here),
        })

res_df = pd.DataFrame(results)

# FDR across ALL tested pairs in this targeted table
res_df["q_fdr"] = fdr_bh(res_df["p"].to_numpy())

# Create pivot tables
R = res_df.pivot(index="metric", columns="target", values="partial_r")
Q = res_df.pivot(index="metric", columns="target", values="q_fdr")
P = res_df.pivot(index="metric", columns="target", values="p")

display(R)
display(Q)
```

#  Jonckheere–Terpstra test

```python
def jonckheere_terpstra(df, metric, group_col="expertise_B_ord"):
    """
    Tests whether metric shows a monotonic trend across ordered groups.
    Returns statistic J and p-value.
    """
    groups = []
    for g in sorted(df[group_col].dropna().unique()):
        vals = df.loc[df[group_col] == g, metric].dropna().values
        if len(vals) > 0:
            groups.append(vals)

    if len(groups) < 3:
        return np.nan, np.nan

    J, p = stats.jonckheere_terpstra(groups)
    return J, p
```

```python
ORDINAL_METRICS = [
    "align_lag_asymmetry_AminusB",
    "centroid_dist_slope",
    "entropy_total",
    "containment_A_in_B",
    "containment_B_in_A",
    "hull_overlap_ratio_min",
]
```

```python
rows = []
for m in ORDINAL_METRICS:
    if m not in df.columns:
        continue

    J, p_jt = jonckheere_terpstra(df, m)
    rho, p_sp = stats.spearmanr(df["expertise_B_ord"], df[m], nan_policy="omit")

    rows.append({
        "metric": m,
        "JT_stat": J,
        "JT_p": p_jt,
        "spearman_rho": rho,
        "spearman_p": p_sp,
        "direction": "increasing" if rho > 0 else "decreasing"
    })

ordinal_results = pd.DataFrame(rows).sort_values("JT_p")
ordinal_results
```

# Effect size + consistency analysis (pattern stability)

```python
def cliffs_delta(x, y):
    x = np.asarray(x)
    y = np.asarray(y)
    nx, ny = len(x), len(y)
    if nx == 0 or ny == 0:
        return np.nan

    gt = sum(a > b for a in x for b in y)
    lt = sum(a < b for a in x for b in y)
    return (gt - lt) / (nx * ny)
```

```python
def pairwise_effects(df, metric):
    results = []
    levels = sorted(df["expertise_B_ord"].unique())

    for i in range(len(levels)):
        for j in range(i+1, len(levels)):
            g1, g2 = levels[i], levels[j]
            x = df.loc[df["expertise_B_ord"] == g1, metric].dropna()
            y = df.loc[df["expertise_B_ord"] == g2, metric].dropna()

            delta = cliffs_delta(x, y)
            results.append({
                "metric": metric,
                "group_1": g1,
                "group_2": g2,
                "cliffs_delta": delta
            })
    return pd.DataFrame(results)
```

```python
pairwise_effects(df, "centroid_dist_mean")
```

```python
def bootstrap_spearman(df, metric, n_boot=2000):
    rhos = []
    for _ in range(n_boot):
        samp = df.sample(frac=1, replace=True)
        rho, _ = stats.spearmanr(samp["expertise_B_ord"], samp[metric], nan_policy="omit")
        rhos.append(rho)
    return np.array(rhos)
```

```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

# -------------------------
# Bootstrap Spearman for ALL metrics
# -------------------------
def _safe_numeric(series: pd.Series) -> pd.Series:
    return pd.to_numeric(series, errors="coerce")

def bootstrap_spearman_series(x, y, n_boot=2000, seed=0):
    """
    Bootstraps Spearman rho between x (ordinal) and y (metric).
    Returns rhos (np.array) after dropping NaNs pairwise each resample.
    """
    rng = np.random.default_rng(seed)
    x = _safe_numeric(pd.Series(x))
    y = _safe_numeric(pd.Series(y))

    # base paired non-NaN mask
    base = pd.DataFrame({"x": x, "y": y}).dropna()
    n = len(base)
    if n < 8:
        return np.array([])

    rhos = np.empty(n_boot, dtype=float)
    for b in range(n_boot):
        idx = rng.integers(0, n, size=n)  # resample rows
        samp = base.iloc[idx]
        rho, _ = stats.spearmanr(samp["x"], samp["y"])
        rhos[b] = rho
    return rhos

def bootstrap_all_metrics(
    df: pd.DataFrame,
    *,
    x_col="expertise_B_ord",
    metrics=None,
    n_boot=2000,
    seed=0,
    min_n=8,
):
    """
    Returns:
      summary_df: one row per metric with mean/median/CI + two-sided bootstrap p-value
      rhos_dict:  {metric: np.array of bootstrapped rhos}
    """
    if metrics is None:
        # default: all numeric columns except obvious non-metrics
        exclude = {x_col, "expertise_gap_abs", "expertise_gap_signed"}
        metrics = [
            c for c in df.columns
            if c not in exclude and pd.api.types.is_numeric_dtype(df[c])
        ]

    rhos_dict = {}
    rows = []
    for m in metrics:
        if m not in df.columns:
            continue

        # paired valid n for this metric
        d = df[[x_col, m]].copy()
        d[x_col] = _safe_numeric(d[x_col])
        d[m] = _safe_numeric(d[m])
        d = d.dropna()
        n = len(d)
        if n < max(min_n, 8):
            continue

        rhos = bootstrap_spearman_series(d[x_col], d[m], n_boot=n_boot, seed=seed)
        if rhos.size == 0:
            continue

        mean_r = float(np.nanmean(rhos))
        med_r  = float(np.nanmedian(rhos))
        lo, hi = np.nanpercentile(rhos, [2.5, 97.5])

        # two-sided "bootstrap p": how often rho is on the opposite side of 0
        # (equivalently, 2*min(P(rho<=0), P(rho>=0)))
        p_boot = 2 * min(np.mean(rhos <= 0), np.mean(rhos >= 0))
        p_boot = float(min(1.0, p_boot))

        rows.append({
            "metric": m,
            "n_pairs": int(n),
            "rho_mean": mean_r,
            "rho_median": med_r,
            "rho_ci2p5": float(lo),
            "rho_ci97p5": float(hi),
            "p_boot_2sided": p_boot,
        })
        rhos_dict[m] = rhos

    summary_df = pd.DataFrame(rows).sort_values("rho_mean", ascending=True)

    return summary_df, rhos_dict

# -------------------------
# Optional: BH-FDR correction for the bootstrap p-values
# -------------------------
def fdr_bh(pvals: np.ndarray):
    p = np.asarray(pvals, dtype=float)
    q = np.full_like(p, np.nan)
    mask = ~np.isnan(p)
    pv = p[mask]
    m = pv.size
    if m == 0:
        return q
    order = np.argsort(pv)
    ranked = pv[order]
    q_ranked = ranked * m / (np.arange(1, m + 1))
    q_ranked = np.minimum.accumulate(q_ranked[::-1])[::-1]
    q_ranked = np.clip(q_ranked, 0, 1)
    q_tmp = np.empty_like(pv)
    q_tmp[order] = q_ranked
    q[mask] = q_tmp
    return q

# -------------------------
# Optional: plotting helpers
# -------------------------
def plot_bootstrap_hist(rhos, title="", bins=40):
    plt.figure(figsize=(6,4))
    plt.hist(rhos, bins=bins)
    plt.axvline(np.nanmean(rhos), linestyle="--")
    plt.axvline(0, linestyle=":", linewidth=1)
    plt.title(title)
    plt.xlabel("Spearman rho")
    plt.ylabel("count")
    plt.tight_layout()
    plt.show()

def plot_top_k_bootstraps(rhos_dict, summary_df, k=6, order_by="abs"):
    d = summary_df.copy()
    d["abs_r"] = d["rho_mean"].abs()
    if order_by == "abs":
        pick = d.sort_values("abs_r", ascending=False).head(k)["metric"].tolist()
    else:
        pick = d.sort_values("rho_mean", ascending=(order_by=="neg")).head(k)["metric"].tolist()

    for m in pick:
        plot_bootstrap_hist(rhos_dict[m], title=f"{m} | bootstrap rho")

# -------------------------
# USAGE
# -------------------------
# Choose which metrics to bootstrap
METRICS = [
    "centroid_dist_mean",
    "centroid_dist_slope",
    "centroid_dist_std",
    "cross_nn_A_to_B",
    "cross_nn_B_to_A",
    "hausdorff_AB",
    "turn_jump_mean",
    "entropy_total",
    "entropy_A",
    "entropy_B",
    "entropy_slope_A",
    "entropy_slope_B",
    "mi_speaker_cluster",
    "nmi_speaker_cluster",
    "mi_turn_clusters",
    "nmi_turn_clusters",
    "hull_overlap_ratio_min",
    "containment_A_in_B",
    "containment_B_in_A",
    "align_lag_asymmetry_AminusB",
    "influence_asymmetry_AminusB",
]

# Run
summary_df, rhos_dict = bootstrap_all_metrics(
    df,
    x_col="expertise_B_ord",
    metrics=[m for m in METRICS if m in df.columns],
    n_boot=5000,   # bump if you want smoother CIs
    seed=42,
    min_n=8,
)

# Add FDR q-values across metrics (optional)
summary_df["q_fdr"] = fdr_bh(summary_df["p_boot_2sided"].to_numpy())

display(summary_df.sort_values("q_fdr").head(20))

# Plot a few strongest effects (by |mean rho|)
plot_top_k_bootstraps(rhos_dict, summary_df, k=6, order_by="abs")
```
